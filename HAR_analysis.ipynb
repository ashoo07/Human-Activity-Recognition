```python
all_dfs = []

for f in sorted(os.listdir(folder)):
    if f.endswith(".arff"):
        data, meta = arff.loadarff(os.path.join(folder, f))
        d = pd.DataFrame(data)

        # same cleaning as before
        d.columns = d.columns.str.replace('"', '', regex=False)
        for col in d.columns:
            if d[col].dtype == object:
                d[col] = d[col].apply(
                    lambda v: v.decode("utf-8") if isinstance(v, bytes) else v
                )
                d[col] = d[col].astype(str).str.replace('"', '', regex=False)

        all_dfs.append(d)

full = pd.concat(all_dfs, ignore_index=True)   # stack all tables together
print(full.shape)
print("People:", full["class"].nunique())
```

```python
def load_sensor(folder_path, tag):
    dfs = []
    for f in sorted(os.listdir(folder_path)):
        if f.endswith(".arff"):
            data, meta = arff.loadarff(os.path.join(folder_path, f))
            d = pd.DataFrame(data)

            d.columns = d.columns.str.replace('"', '', regex=False)
            for col in d.columns:
                if d[col].dtype == object:
                    d[col] = d[col].apply(
                        lambda v: v.decode("utf-8") if isinstance(v, bytes) else v
                    )
                    d[col] = d[col].astype(str).str.replace('"', '', regex=False)

            d["subject"] = f.split("_")[1]
            d["win"] = d.groupby("ACTIVITY").cumcount()   # window number inside each activity
            dfs.append(d)

    out = pd.concat(dfs, ignore_index=True).drop(columns=["class"])

    keys = ["ACTIVITY", "subject", "win"]
    out = out.rename(columns={c: f"{tag}_{c}" for c in out.columns if c not in keys})
    return out
```

```python
pa = load_sensor(r"C:\HAR\phone_accel", "pa")
pg = load_sensor(r"C:\HAR\phone_gyro",  "pg")
wa = load_sensor(r"C:\HAR\watch_accel", "wa")
wg = load_sensor(r"C:\HAR\watch_gyro",  "wg")

for name, t in [("phone accel", pa), ("phone gyro", pg), ("watch accel", wa), ("watch gyro", wg)]:
    print(name, t.shape, "people:", t["subject"].nunique())
```

```python
keys = ["ACTIVITY", "subject", "win"]

merged = (pa.merge(pg, on=keys)
            .merge(wa, on=keys)
            .merge(wg, on=keys))

print(merged.shape)
print("People:", merged["subject"].nunique())
```

```python
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.model_selection import GroupKFold, cross_val_score

y_m = merged["ACTIVITY"]
groups_m = merged["subject"]
gkf = GroupKFold(n_splits=5)

combos = {
    "Phone accel only":      ["pa_"],
    "Phone accel + gyro":    ["pa_", "pg_"],
    "Watch accel + gyro":    ["wa_", "wg_"],
    "All four sensors":      ["pa_", "pg_", "wa_", "wg_"],
}

for name, prefixes in combos.items():
    cols = [c for c in merged.columns if c.startswith(tuple(prefixes))]
    X_m = merged[cols].astype(float)
    et = ExtraTreesClassifier(n_estimators=200, random_state=42, n_jobs=-1)
    s = cross_val_score(et, X_m, y_m, cv=gkf, groups=groups_m)
    print(f"{name}: {s.mean():.3f}  ({len(cols)} features)")
```

```python
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import confusion_matrix

cols_all = [c for c in merged.columns if c not in ["ACTIVITY", "subject", "win"]]
X_all = merged[cols_all].astype(float)

et = ExtraTreesClassifier(n_estimators=200, random_state=42, n_jobs=-1)
y_pred_all = cross_val_predict(et, X_all, y_m, cv=gkf, groups=groups_m)

labels = sorted(y_m.unique())
names = [activity_names[l] for l in labels]
cm = confusion_matrix(y_m, y_pred_all, labels=labels, normalize="true")

plt.figure(figsize=(12, 10))
sns.heatmap(cm, annot=True, fmt=".2f", cmap="Blues",
            xticklabels=names, yticklabels=names)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix: all 4 sensors, 5-fold by person")
plt.tight_layout()
plt.savefig(r"C:\HAR\confusion_matrix.png", dpi=150)
plt.show()
```

```python
et = ExtraTreesClassifier(n_estimators=200, random_state=42, n_jobs=-1)
et.fit(X_all, y_m)

imp = pd.Series(et.feature_importances_, index=cols_all)
by_sensor = imp.groupby(lambda c: c.split("_")[0]).sum()
by_sensor.index = by_sensor.index.map(
    {"pa": "Phone accel", "pg": "Phone gyro", "wa": "Watch accel", "wg": "Watch gyro"}
)

by_sensor.sort_values().plot(kind="barh", figsize=(7, 3))
plt.title("Importance by sensor")
plt.xlabel("Share of total importance")
plt.show()

print(imp.sort_values(ascending=False).head(10))
```

```python
from sklearn.model_selection import GroupShuffleSplit

gss = GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42)
tr_idx, te_idx = next(gss.split(X_all, y_m, groups_m))

X_tr, X_te = X_all.iloc[tr_idx], X_all.iloc[te_idx]
y_tr, y_te = y_m.iloc[tr_idx], y_m.iloc[te_idx]
g_tr = groups_m.iloc[tr_idx]

print("Tune on people:", g_tr.nunique())
print("Final test people:", groups_m.iloc[te_idx].nunique())
```

```python
from sklearn.metrics import accuracy_score

base = ExtraTreesClassifier(n_estimators=200, random_state=42, n_jobs=-1)
base.fit(X_tr, y_tr)
base_acc = accuracy_score(y_te, base.predict(X_te))
print("Baseline accuracy on held-out people:", round(base_acc, 3))
```

```python
import time
from sklearn.model_selection import RandomizedSearchCV, GroupKFold

param_grid = {
    "n_estimators": [200, 400],
    "max_features": ["sqrt", "log2", 0.1, 0.2],
    "min_samples_leaf": [1, 2, 3, 5],
    "max_depth": [None, 20, 40],
    "class_weight": [None, "balanced"],
}

search = RandomizedSearchCV(
    ExtraTreesClassifier(random_state=42, n_jobs=1),
    param_distributions=param_grid,
    n_iter=8,
    cv=GroupKFold(n_splits=4),
    scoring="accuracy",
    random_state=42,
    n_jobs=-1,
    verbose=2,
)

start = time.time()
search.fit(X_tr, y_tr, groups=g_tr)
print("Time (minutes):", round((time.time() - start) / 60, 1))
print("Best settings:", search.best_params_)
print("Best CV score (tuning people only):", round(search.best_score_, 3))
```

```python
best = search.best_estimator_
tuned_acc = accuracy_score(y_te, best.predict(X_te))

print("Baseline:", round(base_acc, 3))
print("Tuned:   ", round(tuned_acc, 3))
print("Change:  ", round(tuned_acc - base_acc, 3))
```
